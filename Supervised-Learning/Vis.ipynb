# Matplotlib Chart Examples

This notebook demonstrates many common Matplotlib chart types: line, scatter, bar, pie, histogram, box/violin, stackplot, heatmap, contour, quiver, polar, hexbin, and 3D plots. Run the code cell below to render examples.

```python
# Not nessarily need to import matplotlib as already imported above
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

categories = ['A', 'B', 'C', 'D']
values = [15, 25, 35, 25]

data = [10, 20, 25, 30, 40, 40, 50, 60, 70, 80, 90, 100]

# creating subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# line chart
axs[0, 0].plot(x, y, marker='o', color='blue')
axs[0, 0].set_title("Line Chart")
axs[0, 0].set_xlabel("X Axis")
axs[0, 0].set_ylabel("Y Axis")

# bar chart
axs[0, 1].bar(categories, values, color='green')
axs[0, 1].set_title("Bar Chart")
axs[0, 1].set_xlabel("Categories")
axs[0, 1].set_ylabel("Values")

# 3️⃣ Pie Chart
axs[1, 0].pie(values, labels=categories, autopct='%1.1f%%', startangle=90)
axs[1, 0].set_title("Pie Chart")


# 4️⃣ Histogram
axs[1, 1].hist(data, bins=6, color='orange', edgecolor='black')
axs[1, 1].set_title("Histogram")
axs[1, 1].set_xlabel("Data Range")
axs[1, 1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
